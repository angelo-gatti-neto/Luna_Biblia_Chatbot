# 🙏 Projeto: Luna Bíblia Chatbot

Este projeto consiste na criação de um chatbot chamado **Luna**, especializado em **conhecimento bíblico**, utilizando a API da Grop com o modelo **llama3-70b-8192**.

A Luna responde perguntas sobre a Bíblia, cita versículos, compartilha ensinamentos cristãos e oferece mensagens de fé de forma acolhedora, respeitosa e inspiradora.

Agora com a opção de escolher entre diferentes categorias de conteúdo:

- Pergunta geral
- Devocional diário
- Histórias bíblicas
- Ensinamentos de Jesus
- Versículos
- Princípios cristãos

Este projeto foi desenvolvido para demonstrar a integração de **Inteligência Artificial** com **temas religiosos**, focando em aplicações práticas e positivas de IA.

---

**Tecnologias utilizadas:**
- Python 3.8+
- Grop API (modelo llama3-70b-8192)
- Jupyter Notebook


## 📦 Instalação de Biblioteca

## 🔑 Configurar a chave da API Groq

Configure a sua chave para autenticar as requisições à Groq API.
Usamos um arquivo `config.py` (não enviado ao GitHub) para manter a chave segura.

```python
# Importar a chave da Groq API do arquivo config.py
from config import API_KEY

from groq import Groq

# Criar o cliente Groq
client = Groq(api_key=API_KEY)

```

## 🧠 Definir o Prompt Base da Luna Bíblia

Aqui configuramos a personalidade da assistente Luna usando um prompt especial.
O prompt molda a forma como a IA irá responder, focando em ensinamentos bíblicos e mensagens de fé.


```python
# Definição do comportamento da Luna
PROMPT_BASE = """
Você é Luna, uma assistente cristã especializada em conhecimento bíblico. 
Sua missão é ajudar com amor, sabedoria e humildade, sempre citando passagens da Bíblia 
e incentivando a fé em Deus. Quando não souber a resposta, seja sincera e acolhedora.
"""

```

## 🤖 Criar a função de geração de resposta personalizada

Agora vamos criar uma função que envia a pergunta do usuário para o Grop API
e retorna a resposta da Luna Bíblia.


```python
# Função para gerar a resposta da Luna conforme a categoria
def gerar_resposta_categoria(pergunta_usuario, categoria):
    if categoria == "Pergunta":
        prompt_categoria = PROMPT_BASE
    elif categoria == "Devocional diário":
        prompt_categoria = PROMPT_BASE + "\nOfereça um devocional diário motivador e reconfortante."
    elif categoria == "Histórias bíblicas":
        prompt_categoria = PROMPT_BASE + "\nConte uma história bíblica interessante e com ensinamento."
    elif categoria == "Ensinamentos de Jesus":
        prompt_categoria = PROMPT_BASE + "\nFoque nos ensinamentos de Jesus para orientar e inspirar."
    elif categoria == "Versículos":
        prompt_categoria = PROMPT_BASE + "\nForneça um versículo bíblico com contexto explicativo."
    elif categoria == "Princípios cristãos":
        prompt_categoria = PROMPT_BASE + "\nExplique um princípio cristão fundamental de forma clara e inspiradora."
    else:
        prompt_categoria = PROMPT_BASE

    resposta = client.chat.completions.create(
        model="llama3-70b-8192",
        messages=[
            {"role": "system", "content": prompt_categoria},
            {"role": "user", "content": pergunta_usuario}
        ],
        temperature=0.7,
        max_tokens=700
    )
    return resposta.choices[0].message.content
```

## 🗣️ Interação com a Luna Bíblia e o Menu Interativo de Seleção de Categoria

Agora vamos conversar com a Luna.  
Digite sua pergunta sobre a Bíblia e receba uma resposta inspiradora!


```python
# Menu de interação
print("=== Escolha a Categoria ===")
print("1 - Pergunta Geral")
print("2 - Devocional Diário")
print("3 - Histórias Bíblicas")
print("4 - Ensinamentos de Jesus")
print("5 - Versículos")
print("6 - Princípios Cristãos")
opcao = input("Digite o número da categoria desejada: ")

opcoes = {
    "1": "Pergunta",
    "2": "Devocional diário",
    "3": "Histórias bíblicas",
    "4": "Ensinamentos de Jesus",
    "5": "Versículos",
    "6": "Princípios cristãos"
}

categoria = opcoes.get(opcao, "Pergunta")
pergunta = input("Digite sua pergunta ou deixe vazio para receber conteúdo automático: ")
if not pergunta.strip():
    pergunta = "Forneça um conteúdo relevante para a categoria selecionada."

# Gerar e mostrar a resposta
resposta = gerar_resposta_categoria(pergunta, categoria)
print("\nLuna Bíblia:")
print(resposta)
```
